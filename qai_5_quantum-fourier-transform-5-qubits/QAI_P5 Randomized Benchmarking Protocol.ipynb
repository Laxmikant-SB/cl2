```python
# Import required libraries
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit, transpile
from qiskit.circuit.library import QFT
from qiskit_aer import AerSimulator

# Create simulator
simulator = AerSimulator()

# ==========================================
# PART 1: 5-Qubit Quantum Fourier Transform
# ==========================================

q = QuantumRegister(5, 'q')
c = ClassicalRegister(5, 'c')

circuit = QuantumCircuit(q, c)

# Prepare input state |10101>
circuit.x(q[4])
circuit.x(q[2])
circuit.x(q[0])

circuit.barrier()

# Apply 5-qubit QFT
qft = QFT(
    num_qubits=5,
    approximation_degree=0,
    do_swaps=True,
    inverse=False,
    insert_barriers=False,
    name='qft'
)

circuit.append(qft, q)

# Measure all qubits
circuit.measure(q, c)

# Display circuit
print("5-Qubit QFT Circuit:")
print(circuit.draw())

# Execute circuit
compiled_circuit = transpile(circuit, simulator)
result = simulator.run(compiled_circuit, shots=1000).result()

counts = result.get_counts()

print("\nQFT Output")
print("-----------")
print(counts)


# ==========================================
# PART 2: QFT WITH INVERSE QFT
# ==========================================

q = QuantumRegister(5, 'q')
c = ClassicalRegister(5, 'c')

circuit2 = QuantumCircuit(q, c)

# Prepare input state |10101>
circuit2.x(q[4])
circuit2.x(q[2])
circuit2.x(q[0])

circuit2.barrier()

# Apply QFT
qft = QFT(
    num_qubits=5,
    approximation_degree=0,
    do_swaps=True,
    inverse=False,
    insert_barriers=False,
    name='qft'
)

circuit2.append(qft, q)

circuit2.barrier()

# Apply Inverse QFT
inverse_qft = QFT(
    num_qubits=5,
    approximation_degree=0,
    do_swaps=True,
    inverse=True,
    insert_barriers=False,
    name='inverse_qft'
)

circuit2.append(inverse_qft, q)

# Measure all qubits
circuit2.measure(q, c)

# Display circuit
print("\nQFT with Inverse QFT Circuit:")
print(circuit2.draw())

# Execute circuit
compiled_circuit2 = transpile(circuit2, simulator)
result2 = simulator.run(compiled_circuit2, shots=1000).result()

counts2 = result2.get_counts()

print("\nQFT with Inverse QFT Output")
print("----------------------------")
print(counts2)
```

## Oral Questions and Answers

### Q1. What is Quantum Computing?
Quantum computing is a type of computing that uses quantum bits (qubits) to perform calculations.

### Q2. What is a qubit?
A qubit is the basic unit of quantum information. It can exist in state 0, state 1, or a superposition of both.

### Q3. What is Quantum Fourier Transform (QFT)?
QFT is a quantum circuit that transforms a quantum state from the computational basis to the Fourier basis.

### Q4. What is the purpose of QFT?
QFT is used to transform quantum states and is an important part of quantum algorithms such as Shor's algorithm.

### Q5. What is the computational basis?
The computational basis consists of the basic states |0> and |1> of a qubit.

### Q6. What is the Fourier basis?
The Fourier basis represents quantum states using different phase relationships and superposition.

### Q7. What is superposition?
Superposition is the ability of a qubit to exist in a combination of |0> and |1> states until measurement.

### Q8. Which gates are used to implement QFT?
QFT uses Hadamard gates, controlled phase rotation gates, and SWAP gates.

### Q9. What is a Hadamard gate?
A Hadamard gate creates superposition from a computational basis state.

### Q10. What is a Controlled U1 gate?
A Controlled U1 gate applies a phase rotation to the target qubit when the control qubit is 1.

### Q11. What is the role of a SWAP gate in QFT?
A SWAP gate exchanges the states of two qubits. It is used to reverse the order of qubits in the QFT circuit.

### Q12. What is the purpose of a 5-qubit QFT?
It transforms the state of five qubits from the computational basis to the Fourier basis.

### Q13. What is the role of the X gate?
The X gate flips a qubit from |0> to |1> or from |1> to |0>.

### Q14. Why are X gates applied to q[4], q[2], and q[0]?
They prepare the input state |10101> before applying QFT.

### Q15. What is the purpose of the QuantumRegister?
QuantumRegister creates and stores quantum bits in a quantum circuit.

### Q16. What is the purpose of the ClassicalRegister?
ClassicalRegister stores the measurement results of qubits as classical bits.

### Q17. What is QuantumCircuit?
QuantumCircuit is a Qiskit class used to create quantum circuits and apply quantum gates.

### Q18. What is the purpose of the QFT() function?
The QFT() function creates a Quantum Fourier Transform circuit for the specified number of qubits.

### Q19. What does num_qubits=5 mean?
It specifies that the QFT circuit will operate on five qubits.

### Q20. What is approximation_degree?
It determines how many small-angle rotations can be ignored to reduce circuit depth. A value of 0 means no approximation.

### Q21. What does do_swaps=True mean?
It includes SWAP gates to arrange the output qubits in the correct order.

### Q22. What does inverse=False mean?
It means that the normal Quantum Fourier Transform is applied.

### Q23. What does inverse=True mean?
It means that the Inverse Quantum Fourier Transform is applied.

### Q24. What is Inverse Quantum Fourier Transform (IQFT)?
IQFT is the reverse operation of QFT. It transforms the Fourier basis back into the computational basis.

### Q25. Why do we use inverse QFT?
We use inverse QFT to recover the original quantum state after applying QFT.

### Q26. What happens when QFT is followed by inverse QFT?
The original quantum state is recovered.

### Q27. What is a quantum measurement?
Measurement converts the quantum state into a classical result, either 0 or 1 for each qubit.

### Q28. Why do we measure the qubits?
We measure the qubits to obtain the final classical output.

### Q29. What is the purpose of the barrier() gate?
A barrier separates circuit operations and makes the circuit easier to understand.

### Q30. What is AerSimulator?
AerSimulator is a Qiskit simulator used to execute quantum circuits on a local computer.

### Q31. What is transpile()?
Transpile() converts a quantum circuit into instructions supported by the selected backend.

### Q32. What is shots=1000?
It means the quantum circuit is executed 1000 times to collect measurement results.

### Q33. What is get_counts()?
get_counts() returns the number of times each measurement result occurs.

### Q34. Why do we get multiple outputs after applying QFT?
QFT creates a superposition of states, so measurement can produce different binary outcomes.

### Q35. What is the expected output after QFT followed by inverse QFT?
The original input state is recovered. In this practical, the expected output is 10101.

### Q36. What is the difference between QFT and inverse QFT?

| QFT | Inverse QFT |
|---|---|
| Transforms computational basis into Fourier basis. | Transforms Fourier basis back into computational basis. |
| Uses forward phase rotations. | Uses reversed phase rotations. |
| inverse=False | inverse=True |

### Q37. What is the importance of QFT in quantum algorithms?
QFT is an important operation used in quantum algorithms such as Shor's factoring algorithm and quantum phase estimation.

### Q38. What is quantum phase estimation?
Quantum phase estimation is a quantum algorithm used to estimate the phase associated with an eigenvalue of a unitary operator. It uses inverse QFT.

### Q39. What is the difference between a classical bit and a qubit?

| Classical Bit | Qubit |
