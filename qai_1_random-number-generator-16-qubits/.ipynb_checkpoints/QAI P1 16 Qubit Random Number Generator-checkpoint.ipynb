```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

# Create 16-qubit circuit
qc = QuantumCircuit(16, 16)

# Apply Hadamard gate to all qubits
qc.h(range(16))

# Measure all qubits
qc.measure(range(16), range(16))

# Run simulator
simulator = AerSimulator()
compiled = transpile(qc, simulator)
result = simulator.run(compiled, shots=1).result()

counts = result.get_counts()

random_number = list(counts.keys())[0]

print("16-Bit Quantum Random Number:")
print(random_number)

print("\nCircuit:")
print(qc.draw())
```

### Q3. What are the different types of registers?
There are two main types:
- Quantum Register: Stores qubits.
- Classical Register: Stores measurement results as classical bits.

### Q4. What is a Quantum Circuit?
A quantum circuit is a sequence of quantum gates and operations performed on qubits to execute a quantum algorithm.

### Q5. How do you measure qubits?
We use the measurement operation to convert the quantum state into a classical result (0 or 1).

### Q6. What is a Hadamard Gate?
A Hadamard gate creates an equal superposition of 0 and 1 when applied to the |0⟩ state.

It is represented by H.
