```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Ensure NLTK resources are available
nltk.download('stopwords')
nltk.download('punkt') # Added to download the 'punkt' tokenizer data
nltk.download('punkt_tab') # Added to download the 'punkt_tab' resource as specified by the error

# 1. Create a sample file as instructed
with open('file.txt', 'w', encoding='utf8') as f:
    f.write("This is the first word.\n")
    f.write("This is the second text, Hello! How are you?\n")
    f.write("This is the third, this is it now.\n")

# 2. Read and process the file
with open('file.txt', 'r', encoding='utf8') as file:
    lines = file.readlines()

stop_words = set(stopwords.words('english'))
inverted_index = {}

# 3. Build the Inverted Index
for line_num, line in enumerate(lines, 1):
    # Remove punctuation and convert to lowercase
    clean_line = "".join([char for char in line.lower() if char.isalnum() or char.isspace()])

    # Tokenize and remove stop words
    tokens = word_tokenize(clean_line)
    tokens_without_sw = [word for word in tokens if word not in stop_words]

    for word in tokens_without_sw:
        if word not in inverted_index:
            inverted_index[word] = []
        if line_num not in inverted_index[word]:
            inverted_index[word].append(line_num)

print("Inverted Index:\n", inverted_index)
```


# IR Practical 2 – Oral / Viva Questions and Answers

### Q1. What is an Inverted Index?
An Inverted Index is a data structure that stores each word along with the list of documents or lines in which that word appears.

### Q2. Why do we use an Inverted Index?
It helps us search and retrieve documents quickly without checking every document word by word.

### Q3. What are the main components of an Inverted Index?
1. Term (word)
2. Posting List (list of document IDs or line numbers)

### Q4. What is a Posting List?
A Posting List contains the document IDs or line numbers where a particular word appears.

### Q5. What is the difference between a normal index and an inverted index?
- Normal index: Maps documents to words.
- Inverted index: Maps words to documents.

### Q6. Which data structure is used in your program?
A Python dictionary is used to create the inverted index.

### Q7. What is tokenization?
Tokenization is the process of splitting text into smaller units called tokens.

To maintain uniformity and avoid treating uppercase and lowercase words as different words.

Stop words are commonly occurring words such as "is", "the", "and", and "in" that are usually removed during text preprocessing.

### Q10. Why do we remove punctuation?
To avoid treating words with punctuation as different terms.

### Q11. Which library is used for tokenization?
